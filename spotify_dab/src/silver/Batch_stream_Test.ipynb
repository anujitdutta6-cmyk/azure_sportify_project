```python
from pyspark.sql.functions import col, when

# 1. Temporarily use spark.read (Batch) instead of spark.readStream (Stream)
df_track_test = spark.read.format("parquet")\
    .load("abfss://bronze@anujspotifyazureproject.dfs.core.windows.net/DimTrack") # Update path if needed

# 2. Apply your Transformation
df_track_test = df_track_test.withColumn("durationFlag", 
                                         when(col('duration_sec') < 150, "low")\
                                         .when(col('duration_sec') < 300, "medium")\
                                         .otherwise("high"))

# 3. Display works instantly without any checkpoint errors!
display(df_track_test)
```

```python
from pyspark.sql.functions import col, when, regexp_replace

# 1. READ (Batch Mode for Testing)
df_track_test = (spark.read.format("parquet")
    .load("abfss://bronze@anujspotifyazureproject.dfs.core.windows.net/DimTrack")
)

# 2. APPLY TRANSFORMATIONS
df_track_test = (df_track_test
    .withColumn("durationFlag", 
                when(col('duration_sec') < 150, "low")
                .when(col('duration_sec') < 300, "medium")
                .otherwise("high")
    )
    .withColumn("track_name", regexp_replace(col('track_name'), '-', ' '))
)

# 3. PREVIEW RESULTS
display(df_track_test)
```
